```python
import torch
from torch import nn

x = torch.randn(4, 16, 28, 28)
pool = nn.MaxPool2d(2, stride=2)
y = pool(x)

print(y.shape)
```

```python
net = nn.Sequential(
    nn.Conv2d(
        in_channels=3,
        out_channels=8,
        kernel_size=3,
        padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Conv2d(
        in_channels=8,
        out_channels=16,
        kernel_size=3,
        padding=1
    )
)

x = torch.randn(4, 3, 32, 32)

print(net(x).shape)
```
