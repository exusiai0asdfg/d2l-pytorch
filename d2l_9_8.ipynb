```python
import torch
from torch import nn

# 第一步两个 Token：A、B
p_first = torch.tensor([0.6, 0.4])

# 第二步条件概率
# 行对应上一步 A、B
# 列对应下一步 C、D
p_next = torch.tensor([
    [0.51, 0.49],
    [0.95, 0.05]
])

log_p_first = torch.log(p_first).unsqueeze(1)
log_p_next = torch.log(p_next)


joint_log_probs = log_p_first + log_p_next


flat_probs = joint_log_probs.view(-1)
top_probs, top_indices = torch.topk(flat_probs, k=2)

first_indices = top_indices // 2
second_indices = top_indices % 2

print(first_indices, second_indices)
```
