```python
# bidirectional rnn
import torch 
from torch import nn


X = torch.randn(2, 4, 5)

model = nn.LSTM(
    input_size=5,
    hidden_size=8,
    num_layers=2,
    bidirectional=True,
    batch_first=True
)

outputs, (h_last, c_last) = model(X)

print(outputs.shape)
# outputs.shape[1] : four time steps
print(h_last.shape)
# h_last.shape[0] : 2layer*2direction
print(c_last.shape)
 
```

```python
class BiLSTMTagger(nn.Module):
    def __init__(self, input_size, hidden_size, num_classes):
        super().__init__()
        self.model = nn.LSTM(
        input_size=input_size,
        hidden_size=hidden_size,
        num_layers=2,
        bidirectional=True,
        batch_first=True
)
        self.classify = nn.Linear(hidden_size*2,num_classes)

    def forward(self, X):
        outputs, (h_last, c_last) = self.model(X)
        logits = self.classify(outputs)
        return logits
    
model = BiLSTMTagger(5, 8, 3)
X = torch.randn(2, 4, 5)

logits = model(X)
print(logits.shape)
```
