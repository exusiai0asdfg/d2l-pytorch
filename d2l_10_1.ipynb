```python
# attention
import torch
from torch import nn
def simple_attention(q, K, V):
    # q: (D,)
    # K: (N, D)
    # V: (N, Dv)
    s = K@q
    weights = torch.softmax(s,dim=0)
    alpha = weights@V
    output = alpha
    # 1. q 与每个 Key 计算点积分数
    # 2. Softmax 得到 N 个权重
    # 3. 对 N 个 Value 加权求和

    return output, weights

q = torch.tensor([2., 0.])

K = torch.tensor([
    [2., 0.],
    [0., 2.],
    [1., 1.]
])

V = torch.tensor([
    [10., 0.],
    [0., 10.],
    [5., 5.]
])

output, weights = simple_attention(q, K, V)

print(weights)
print(output)
print(weights.sum())
```
