```python
import torch
from torch import nn
class centeredlayer(nn.Module):
    def forward(self,x):
        return x-x.mean()

layer = centeredlayer()
x = torch.randn(4,8)

print(layer(x).shape)
print(layer(x).mean())
```

```python
import torch
from torch import nn
class mylinear(nn.Module):
    def __init__(self):
        super().__init__()
        self.w = nn.Parameter(torch.randn(5,3))
        self.b = nn.Parameter(torch.randn(3))
    def forward(self,x):
        return x@self.w +self.b
net = mylinear()

x = torch.ones(4, 5)
y = net(x)

loss = y.sum()

loss.backward()
print(net.w.grad)
print(net.b.grad)
for name, param in net.named_parameters():
    print(name, param.shape)


```
