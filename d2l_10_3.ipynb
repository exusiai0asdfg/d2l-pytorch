```python
# attention scoring functions 

# additive attention

# scaled dot-product attention
import torch 
from torch import nn
def scaled_dot_product_attention(Q, K, V,valid_lens):
    # Q: (B, Nq, D)
    # K: (B, Nk, D)
    # V: (B, Nk, Dv)
    scores = (Q@K.transpose(-2,-1))/(Q.shape[-1]**0.5)
    B,Nk = K.shape[0],K.shape[1]
    mask = torch.arange(Nk, device=Q.device).unsqueeze(0) >= valid_lens.unsqueeze(1)
    scores = scores.masked_fill(mask.unsqueeze(1), -torch.inf)
    weights = torch.softmax(scores,dim = -1)
    output = weights @ V 
    # 1. 计算每个 Query 与每个 Key 的点积
    # 2. 除以 sqrt(D)
    # 3. 对 Key 维度做 Softmax
    # 4. 用注意力权重对 V 加权求和

    return output, weights


B, Nq, Nk, D, Dv = 2, 3, 5, 4, 6

Q = torch.randn(B, Nq, D)
K = torch.randn(B, Nk, D)
V = torch.randn(B, Nk, Dv)

valid_lens = torch.tensor([3, 4])
```
