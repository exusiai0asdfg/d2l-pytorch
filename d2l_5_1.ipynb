```python
import torch
from torch import nn
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(10,20)
        self.mid = nn.ReLU()
        self.output = nn.Linear(20,2)
    def forward(self,x):
        H = self.mid((self.hidden(x)))
        return self.output(H)
    
net = MLP()
x = torch.randn(5,10)
y = net(x)
print(y.shape)
print(sum(p.numel() for p in net.parameters()))
```

```python
class mysequential(nn.Module):
    def __init__(self,*args):
        super().__init__()
        self.layers = nn.ModuleList(args)
            
    def forward(self,x):
        for layers in self.layers:
            x = layers(x)
        return x
            
x = torch.randn(3,4)
net = mysequential(
    nn.Linear(4, 6),
    nn.ReLU(),
    nn.Linear(6, 2)
)
y = net(x)
print(y.shape)
```

```python
#将多个网络封装到一个普通的列表,pytorch将无法为这几个网络注册,也无法通过api查找变量
class MyNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.layers = [
            nn.Linear(4, 8),
            nn.ReLU(),
            nn.Linear(8, 2)
        ]

    def forward(self, x):
        for layer in self.layers:
            x = layer(x)
        return x
    
net = MyNet()
y = net(x)
print(sum(p.numel() for p in net.parameters()))
```

```python
# 但是可以使用特殊的nn.ModuleList(),其中的网络会被注册
class TestNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.a = nn.Linear(4, 8)
        self.b = [nn.Linear(8, 2)]
        self.c = nn.ModuleList([
            nn.Linear(2, 3)
        ])

    def forward(self, x):
        x = self.a(x)
        x = self.b[0](x)
        x = self.c[0](x)
        return x
    
x = torch.randn(5,4)
net = TestNet()
y = net(x)

print(sum(p.numel() for p in net.parameters()))
```

```python
class Test(nn.Module):
    def __init__(self):
        super().__init__()
        self.w = torch.randn(
            3, 4, requires_grad=True
        )

    def forward(self, x):
        return x @ self.w.T

net = Test()
x = torch.randn(2, 4)
loss = net(x).sum()
loss.backward()
print(net.w.grad)
```
