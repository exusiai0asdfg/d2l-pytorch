```python
import torch
from torch import nn
```

```python
net = nn.Sequential(
    nn.Conv2d(
        in_channels=1,
        out_channels=96,
        kernel_size=11,
        stride=4,
        padding=1
    ),
    nn.ReLU(),
    nn.MaxPool2d(
        kernel_size=3,
        stride=2
    ),
    nn.Conv2d(
        in_channels=96,
        out_channels=256,
        kernel_size=5,
        padding=2,
    ),
    nn.ReLU(),
    nn.MaxPool2d(
        kernel_size=3,
        stride=2
    )
)

x = torch.randn(2,1,224,224)
y = net(x)
```
