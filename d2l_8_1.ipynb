```python
#时间序列->监督学习数据
import torch
from torch import nn

x = torch.arange(1, 11, dtype=torch.float32)
tau = 3

def make_dataset(x, tau):
    features = []
    labels = []

    for i in range(len(x) - tau):
        features.append(x[i:i+tau])
        labels.append(x[i+tau])

    features = torch.stack(features)
    labels = torch.stack(labels).reshape(-1, 1)

    return features, labels

print(make_dataset(x,tau))
```
