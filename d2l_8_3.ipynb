```python
import torch
from torch import nn
```

```python
def make_sequence_batch(corpus, batch_size, num_steps):
    x = []
    y = []
    assert len(corpus)>=batch_size*num_steps+1
    for i in range(batch_size):
        x.append(corpus[0+num_steps*i:num_steps*(i+1)])
        y.append(corpus[0+num_steps*i+1:num_steps*(i+1)+1])
    return torch.stack(x),torch.stack(y)
        
corpus = torch.arange(11)
batch_size = 2
num_steps = 4

X, Y = make_sequence_batch(corpus, batch_size, num_steps)

print(X)
print(Y)
```
