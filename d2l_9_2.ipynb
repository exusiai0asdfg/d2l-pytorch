```python
#LSTM
import torch
from torch import nn
def lstm_update(c_prev, forget, input_gate,
                output_gate, candidate):
    c_next = forget*c_prev + input_gate*candidate
    h_next = output_gate*torch.tanh(c_next)
    return c_next, h_next


c_prev = torch.tensor([[0.8, -0.5]])
forget = torch.tensor([[0.9, 0.8]])
input_gate = torch.tensor([[0.2, 0.1]])
output_gate = torch.tensor([[0.1, 0.9]])
candidate = torch.tensor([[-0.4, 0.6]])

c_next, h_next = lstm_update(
    c_prev, forget, input_gate,
    output_gate, candidate
)

print(c_next)
print(h_next)
```

```python
B, T, D, H = 2, 4, 20, 32

X = torch.randn(B, T, D)

lstm = nn.LSTM(
    input_size=D,
    hidden_size=H,
    num_layers=1,
    batch_first=True
)

outputs, (h_last, c_last) = lstm(X)

print(outputs.shape)
print(h_last.shape)
print(c_last.shape)

print(torch.allclose(
    outputs[:, -1, :],
    h_last[0]
))
```
