```python
#encoder-decoder
import torch
from torch import nn
encoder = nn.LSTM(20, 32, batch_first=True)
decoder = nn.LSTM(20, 32, batch_first=True)

src = torch.randn(4, 7, 20)
tgt_in = torch.randn(4, 5, 20)

outputs,(h_last,c_last) = encoder(src)
de_outputs , _ = decoder(
    tgt_in,(h_last,c_last)
)
print(de_outputs.shape)
```
