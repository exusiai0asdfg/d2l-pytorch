```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

net = nn.Sequential(
    nn.Linear(4,8),
    nn.ReLU(),
    nn.Linear(8,2)
)

net = net.to(device)

x = torch.randn(5,4,device=device)
y = net(x)

print(y.shape)
print(y.device)
print(net[0].weight.device)
```

```python
net = nn.Sequential(
    nn.Linear(6,12),
    nn.ReLU(),
    nn.Linear(12,3)
)
net = net.to(device="cuda")
x = torch.randn(5,6,device="cuda")
y = net(x)
print(y.shape)
print(y.device)
```
