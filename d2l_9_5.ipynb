```python
import torch
def pad_sequence(token_ids, max_len, pad_id=0, eos_id=1):
    if (len(token_ids)<max_len):
        token_ids.append(eos_id)
        for i in range(max_len-len(token_ids)):
            token_ids.append(pad_id)
    else:
        token_ids = token_ids[0:max_len-1]
        token_ids.append(eos_id)
    return torch.tensor(token_ids, dtype=torch.long)
    
    
pad_sequence([5, 8, 3, 7, 9, 4], max_len=4)
## the original token_ids is changed 
```

```python
# more simple one
def pad_sequence(token_ids, max_len, pad_id=0, eos_id=1):
    assert max_len >= 1

    seq = list(token_ids[:max_len - 1])
    seq.append(eos_id)
    seq.extend([pad_id] * (max_len - len(seq)))

    return torch.tensor(seq, dtype=torch.long)

pad_sequence([5, 8, 3, 7, 9, 4], max_len=4)
```
