```python
# Nadaraya–Watson kernel regression
import torch
from torch import nn

def gaussian_attention(q, K, V, sigma):
    # q: 标量
    # K: (N,) 已知输入位置
    # V: (N,) 对应观测值

    s = -(q-K)**2/(2*sigma**2)
    weights = torch.softmax(s,dim=0)
    output = (weights*V).sum()
    # 1. 计算负平方距离分数
    # 2. Softmax
    # 3. 对 V 加权求和

    return output, weights

q = torch.tensor(1.5)
K = torch.tensor([0., 1., 2.])
V = torch.tensor([0., 1., 4.])

for sigma in [0.1, 0.5, 2.0]:
    output, weights = gaussian_attention(q, K, V, sigma)
    print(sigma, output, weights)
```
