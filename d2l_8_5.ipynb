```python
import torch
from torch import nn
def rnn_step(x_t, h_prev, W_xh, W_hh, b_h):
    return torch.tanh(x_t@W_xh+h_prev@W_hh+b_h)
```

```python
def rnn_forward(X, h, W_xh, W_hh, b_h, W_hq, b_q):
    h_prev = h
    out_list = []
    for i in range(X.shape[1]):
        h_next = rnn_step(X[:,i,:],h_prev,W_xh,W_hh,b_h)
        out_t = h_next @ W_hq + b_q
        out_list.append(out_t)
        h_prev = h_next
    outputs = torch.stack(out_list,dim=1)
    return outputs, h_prev

B, T, D, H, Q = 2, 4, 5, 8, 10

X = torch.randn(B, T, D)
h = torch.zeros(B, H)

W_xh = torch.randn(D, H)
W_hh = torch.randn(H, H)
b_h = torch.zeros(H)

W_hq = torch.randn(H, Q)
b_q = torch.zeros(Q)

outputs, h_last = rnn_forward(
    X, h, W_xh, W_hh, b_h, W_hq, b_q
)

print(outputs.shape)
print(h_last.shape)
```
