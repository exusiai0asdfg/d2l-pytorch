```python
# GRU
import torch
from torch import nn
```

```python
def gru_update(h_prev, h_candidate, z):
    return z*h_prev+(1-z)*h_candidate

import torch

h_prev = torch.tensor([[0.8, -0.5, 0.2]])
h_candidate = torch.tensor([[-0.4, 0.9, 0.6]])
z = torch.tensor([[0.9, 0.1, 0.5]])

h_next = gru_update(h_prev, h_candidate, z)

print(h_next)
```

```python
def gru_step(x_t, h_prev,
             W_xz, W_hz, b_z,
             W_xr, W_hr, b_r,
             W_xh, W_hh, b_h):
    z = torch.sigmoid(x_t@W_xz+h_prev@W_hz+b_z)
    r = torch.sigmoid(x_t@W_xr+h_prev@W_hr+b_r)
    h_tidle = torch.tanh(x_t@W_xh+(r*h_prev)@W_hh+b_h)
    h_next = z*h_prev+(1-z)*h_tidle
    return h_next

B, D, H = 2, 5, 8

x_t = torch.randn(B, D)
h_prev = torch.randn(B, H)

# 每个门各有输入权重、隐藏状态权重和偏置
W_xz, W_xr, W_xh = [
    torch.randn(D, H) for _ in range(3)
]

W_hz, W_hr, W_hh = [
    torch.randn(H, H) for _ in range(3)
]

b_z = torch.zeros(H)
b_r = torch.zeros(H)
b_h = torch.zeros(H)

gru_step(x_t, h_prev,
             W_xz, W_hz, b_z,
             W_xr, W_hr, b_r,
             W_xh, W_hh, b_h).shape
```
