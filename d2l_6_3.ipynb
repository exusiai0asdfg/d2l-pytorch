```python
import torch
from torch import nn
conv = nn.Conv2d(
    in_channels=3,
    out_channels=16,
    kernel_size=5,
    padding=2,
    stride=2
)

x = torch.randn(8, 3, 32, 32)
y = conv(x)
print(y.shape)
print(sum(p.numel() for p in conv.parameters()))
```

```python
x = torch.randn(4, 3, 28, 28)
conv1 = nn.Conv2d(
    3, 8, kernel_size=3,
    padding=1, stride=2
)
conv2 = nn.Conv2d(
    8, 16, kernel_size=3,
    padding=0, stride=1
)
y = conv1(x)
print(conv2.weight.shape)
print(sum(p.numel() for p in conv1.parameters()))
print(sum(p.numel() for p in conv2.parameters()))
```

```python
import torch
from torch import nn

x = torch.randn([4,3,28,28])

net = nn.Sequential(
    nn.Conv2d(
    3,8,kernel_size=3,padding=1,stride=2
),
    nn.ReLU(),
    nn.Conv2d(
    8,16,kernel_size=3,padding=0,stride=1
)
)
y_hat = net(x)
print(y_hat.shape)
for name,params in net.named_parameters():
    print(name,params.shape)
```
