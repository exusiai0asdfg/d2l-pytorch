```python
import torch
from torch import nn

def vgg_block(num_convs,in_channels,out_channels):
    layer = []
    now_in = in_channels
    for i in range(num_convs):
        layer.append(nn.Conv2d(in_channels=now_in,out_channels=out_channels,kernel_size=3,padding=1))
        now_in = out_channels
        layer.append(nn.ReLU())
    layer.append(nn.MaxPool2d(kernel_size=2,stride=2))
    
    return nn.Sequential(*layer)

block = vgg_block(
    num_convs=2,
    in_channels=3,
    out_channels=16
)

x = torch.randn(2, 3, 32, 32)
print(block(x).shape)
```

```python
def vgg_block(num_convs,in_channels,out_channels):
    layer = []
    now_in = in_channels
    for i in range(num_convs):
        layer.append(nn.Conv2d(in_channels=now_in,out_channels=out_channels,kernel_size=3,padding=1))
        now_in = out_channels
        layer.append(nn.ReLU())
    layer.append(nn.MaxPool2d(kernel_size=2,stride=2))
    return layer



def vgg(conv_arch):
    layer = []
    in_now = 1
    for tmp in conv_arch:
        num_convs,out_now = tmp
        tmp_layer =vgg_block(num_convs,in_now,out_now)
        layer.extend(tmp_layer)
        in_now = out_now
    layer.append(nn.Flatten())
    layer.append(nn.Linear(3136,128))
    layer.append(nn.ReLU())
    layer.append(nn.Linear(128,10))
    return layer

x = torch.randn(3,1,224,224)
conv_arch = (
    (1, 8),
    (1, 16),
    (2, 32),
    (2, 64),
    (2, 64)
)
layer = vgg(conv_arch)
net = nn.Sequential(*layer)

y = net(x)
print(y.shape)
```

```python
## a better way

def vgg(conv_arch):
    blocks = []
    in_channels = 1

    for num_convs, out_channels in conv_arch:
        blocks.append(
            vgg_block(num_convs, in_channels, out_channels)
        )
        in_channels = out_channels

    return nn.Sequential(
        *blocks,
        nn.Flatten(),
        nn.Linear(in_channels * 7 * 7, 128),
        nn.ReLU(),
        nn.Linear(128, 10)
    )
```
