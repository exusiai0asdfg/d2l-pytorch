```python
from collections import Counter

lines = [
    "the time machine",
    "the time traveller",
    "time moves"
]
```

```python
def build_vocab(lines, min_freq=2):
    tokens = [ ]
    for line in lines:
        tokens.extend(line.split())
    freq = Counter(tokens)
    keep = []
    for w,f in freq.items():
        if(f>=min_freq):
            keep.append((w,f))
    keep.sort(key = lambda x:-x[1])
    vocab = {"<unk>":0}
    for w,f in keep:
        vocab[w] = len(vocab)
    return vocab
    
vocab = build_vocab(lines, min_freq=2)
print(vocab)
```

```python
def encode(sentence,vocab):
    word_list = sentence.split()
    encode_list = []
    for i in word_list:
        if vocab.get(i)==None:
            encode_list.append(0)
        else:
            encode_list.append(vocab.get(i))
    return encode_list
print(encode("the time traveller", vocab))
```
