```python
import torch
from torch import nn

B, T, D, H, Q = 2, 4, 5, 8, 10

X = torch.randn(B, T, D)

rnn = nn.RNN(
    input_size=D,
    hidden_size=H,
    num_layers=1,
    batch_first=True
)

outputs, h_last = rnn(X)

print(outputs.shape)
print(h_last.shape)
```

```python
class RNNLanguageModel(nn.Module):
    def __init__(self, input_size, hidden_size, vocab_size):
        super().__init__()
        self.net1 = nn.RNN(input_size=input_size,hidden_size=hidden_size,batch_first=True)
        self.net2 = nn.Linear(hidden_size,vocab_size)

    def forward(self, X, h=None):
        outputs,h_last = self.net1(X,h)
        logits = self.net2(outputs)
        return logits,h_last       
    
model = RNNLanguageModel(
    input_size=20,
    hidden_size=32,
    vocab_size=20
)

X = torch.randn(4, 6, 20)

logits, h_last = model(X)

print(logits.shape)
print(h_last.shape)
```
